## Benchmark design: matched xCDAT workflows on the same CMIP dataset

xCDAT, a modern CDAT successor built on xarray, streamlines structured-grid climate analysis (e.g., loading, averaging, and regridding).

![Benchmark Design](./images/benchmark-design.png)


![Benchmark Setup](./images/benchmark-setup.png)


<style>img[alt="Workflow-operation timing ratios"] { width: 75%; }</style>



- **Open dominates:** kerchunk won `73/73` cases, with about a `110×` median advantage.
- **Compute crosses over:** temporal and spatial bin medians favor kerchunk starting at `150–199` files.
- **Load remains mixed:** performance varies by file-count bin, without a consistent winner.

> **Takeaway:** Kerchunk provides a durable open-time advantage; operation-level benefits depend on workload and scale.


<style>img[alt="Complete-pipeline timing ratios"] { width: 75%; }</style>



- **Every bin favors kerchunk:** all three pipeline medians remain below the `1×` parity line.
- **Wins are consistent:** kerchunk won `69/73` cases for each tested monthly pipeline.
- **Advantage grows with scale:** every pipeline exceeds a `10×` median advantage from `200–299` files onward.

> **Takeaway:** Kerchunk’s open-time advantage outweighs mixed operation-level performance and drives end-to-end workflow wins.


### Metadata and Output Validation Results:

- **Metadata and CF behavior:** All `21` applicable comparisons passed.
- **xCDAT API outputs:** All `44` completed comparisons passed.
- **Axis detection:** No CF-axis failures.

> **Takeaway:** No regressions were detected between Kerchunk and source NetCDF in tested xCDAT workflows.
