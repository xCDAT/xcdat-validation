# RIOTAI presentation setup

## Environment and Presentation Setup
Run these commands from the repository root. The environment includes xCDAT/RIOTAI dependencies, JupyterLab 4, `jupyterlab_rise`, and `nbconvert`.

```bash
conda env create --file riotai/presentations/environment.yml
conda activate riotai-presentation
jupyter lab riotai/presentations/starter-presentation.ipynb
```

## Slide Editing
In JupyterLab, open the Property Inspector (cog icon at the top right), then use the Common Tools tab to assign each cell a slide type: **Slide**, **Sub-Slide**, **Fragment**, **Skip**, or **Notes**. Start the live presentation with `Ctrl+R` (`Option+R` on macOS).

More information on usage: https://github.com/jupyterlab-contrib/rise#usage

If live presentation mode is unavailable, verify both parts of the extension:

```bash
jupyter server extension list
jupyter labextension list
```

## Presentation Export

Before exporting, run all cells so generated figures are stored as notebook outputs. Export executable Reveal.js slides with embedded Markdown images using:

```bash
jupyter nbconvert \
  --to slides \
  --execute \
  --embed-images \
  --output starter-presentation \
  riotai/presentations/starter-presentation.ipynb
```

This writes `starter-presentation.slides.html`. For portable exports, prefer notebook-generated figures, images attached to Markdown cells, or relative image paths. Avoid absolute paths and remote-only image URLs. Dragging an image into an edited Markdown cell creates an attachment. The example near the end uses an external SVG beside the notebook; `--embed-images` places it inside the exported HTML. The exported slideshow still loads Reveal.js from its configured CDN, so use `--reveal-prefix` with a local Reveal.js copy if the presentation itself must work fully offline.

## Deterministic xarray example

Synthetic data keeps this starter notebook executable without network or HPC filesystem access.

````python
import numpy as np
import xarray as xr

time = np.arange("2026-01", "2027-01", dtype="datetime64[M]")
signal = 0.5 * np.arange(time.size) + np.sin(np.linspace(0, 2 * np.pi, time.size))

ds = xr.Dataset(
    data_vars={"signal": ("time", signal, {"units": "1"})},
    coords={"time": time},
    attrs={"description": "Deterministic starter-presentation data"},
)
ds
````

````python
import matplotlib.pyplot as plt
from IPython.display import display

fig, ax = plt.subplots(figsize=(9, 4.5))
ds["signal"].plot(ax=ax, marker="o", color="#2563eb")
ax.set_title("Generated output stays with the notebook")
ax.grid(alpha=0.25)
fig.tight_layout()
display(fig, metadata={"image/png": {"alt": "Synthetic monthly signal line chart"}})
plt.close(fig)
````

Speaker note: this figure is produced during execution, so no separate PNG path is needed. Mention that `--execute` refreshes it before export.
